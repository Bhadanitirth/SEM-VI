# Aim: To perform KMeans and DBSCAN clustering using sklearn package in python.

```python
# Generate Random Data for clustering
from sklearn.datasets import make_blobs
# from sklearn.datasets.samples_generator import make_blobs
import matplotlib.pyplot as plt
X, y_true = make_blobs(n_samples=100, centers=3, cluster_std=0.6)
plt.scatter(X[:,0],X[:,1], s=50)
```

```python
# Create KMeans Clustering Model
from sklearn.cluster import KMeans
km = KMeans(n_clusters=3)
```

```python
# Run Clustering Algorithm to fit the data and find cluster labels
km.fit(X)
y_lab = km.predict(X)
```

```python
# Plot the data points using scatter-plot
plt.scatter(X[:, 0], X[:, 1], c=y_lab, s=50, cmap='viridis')
centers = km.cluster_centers_
plt.scatter(centers[:, 0], centers[:, 1], c='black', s=200,
alpha=0.5)
```

```python
# Create DBSCAN model
from sklearn.cluster import dbscan
dbs = dbscan(X)
```

```python
# Plot Data Points
plt.scatter(X[:, 0], X[:, 1], c=dbs[1], s=50, cmap='viridis')
```

```python
import pandas as pd
import numpy as np
import sklearn.metrics as sm
import matplotlib.pyplot as plt
from sklearn.cluster import KMeans
import sklearn.datasets as data
```

```python
# Read or Generate the dataset for machine learning task
d1 = data.load_iris()
d1
```

```python
# PLACE THE IRIS DATA IN A PANDAS 
# DATAFRAME 
iris = pd.DataFrame(data=d1.data,columns=d1.feature_names) 
iris
```

```python
# Arrange data into a features matrix (X) and target vector (y)
# Create the Predictor and Target sets
X= d1.data
y= d1.target
```

```python
model=KMeans(n_clusters=3)
```

```python
model.fit(X)
```

```python
y_pred = model.predict(X)
```

```python
y_pred
```

```python
y
```

```python
#Dataframe to show Y real and Y predicted:
#Using a dictionary to separate columns  
result = pd.DataFrame({"Actual": y, "Predicted": y_pred})
result
```

```python
# Compare predicted output with actual output to find accuracy
from sklearn.metrics import classification_report, confusion_matrix, accuracy_score
print(accuracy_score(y,y_pred))
print(confusion_matrix(y,y_pred))
print(classification_report(y, y_pred))
```

```python
# Plot the data points using scatter-plot
plt.scatter(X[:, 0], X[:, 1], c=y_pred, cmap='viridis')
centers = model.cluster_centers_
plt.scatter(centers[:, 0], centers[:, 1], c='black', s=200, alpha=0.5);
```

```python
# Create DBSCAN model
from sklearn.cluster import dbscan
# index, y_pred = dbscan(X)
index , y_pred = dbscan(X,eps=0.5,min_samples=5)
```

```python
#Dataframe to show Y real and Y predicted:
#Using a dictionary to separate columns  
result = pd.DataFrame({"Actual": y, "Predicted": y_pred})
result
```

```python
# Compare predicted output with actual output to find accuracy
from sklearn.metrics import classification_report, confusion_matrix, accuracy_score
print(accuracy_score(y,y_pred))
print(confusion_matrix(y,y_pred))
print(classification_report(y, y_pred))
```
