2 - 2D array creation functions
The 2D array creation functions e.g. numpy.eye, numpy.diag, and numpy.vander define properties of special matrices represented as 2D arrays.

np.eye(n, m) defines a 2D identity matrix. The elements where i=j (row index and column index are equal) are 1 and the rest are 0, as such:

```python
np.eye(3)
```

```python
np.eye(3, 5)
```

numpy.diag can define either a square 2D array with given values along the diagonal or if given a 2D array returns a 1D array that is only the diagonal elements. The two array creation functions can be helpful while doing linear algebra,

```python
np.diag([1, 2, 3])
```

```python
np.diag([1, 2, 3], 1)
```

```python
a = np.array([[1, 2], [3, 4]])
a
```

```python
np.diag(a)
```

vander(x, n) defines a Vandermonde matrix as a 2D NumPy array. Each column of the Vandermonde matrix is a decreasing power of the input 1D array or list or tuple, x where the highest polynomial order is n-1. This array creation routine is helpful in generating linear least squares models,

```python
np.vander(np.linspace(0, 2, 5), 2)
```

```python
np.vander([1, 2, 3, 4], 2)
```

```python
np.vander((1, 2, 3, 4), 4)
```

Replicating, joining, or mutating existing arrays
Once you have created arrays, you can replicate, join, or mutate those existing arrays to create new arrays. When you assign an array or its elements to a new variable, you have to explicitly numpy.copy the array, otherwise the variable is a view into the original array. 

```python
a = np.array([1, 2, 3, 4, 5, 6])
```

```python
b = a[:2]
```

```python
b += 1
```

```python
print('a =', a, '; b =', b)
```

In this example, you did not create a new array. You created a variable, b that viewed the first 2 elements of a. When you added 1 to b you would get the same result by adding 1 to a[:2]. If you want to create a new array, use the numpy.copy array creation routine as such:

```python
a = np.array([1, 2, 3, 4])
b = a[:2].copy()
b += 1
```

```python
print('a = ', a, 'b = ', b)
```

# Categoricals
pandas can include categorical data in a DataFrame.

```python
df = pd.DataFrame( {"id": [1, 2, 3, 4, 5, 6], "raw_grade": ["a", "b", "b", "a", "a", "e"]})
df
```

```python
type(df)
```

```python
df["grade"] = df["raw_grade"].astype("category")
df["grade"]
```

```python
new_categories = ["very good", "good", "very bad"]

df["grade"] = df["grade"].cat.rename_categories(new_categories)
df["grade"]
```

A Parquet file is a column-oriented data file format that's designed to store and retrieve data efficiently: 
Columnar: Parquet stores the values of each table column next to each other, instead of each record. 
Open source: Parquet is free to use and licensed under the Apache Hadoop license. 
Self-describing: Parquet files contain metadata, including the schema and structure, in addition to data. 
Efficient compression: Parquet uses compression and encoding schemes to optimize files for query performance. 
Fast queries: Parquet allows users to fetch specific column values without reading the entire row of data. 
Compatible with many systems: Parquet is supported by many data processing systems, including Spark SQL. 
Parquet files are well suited for handling large volumes of complex data. They're more efficient than other unstructured data formats, like CSV or JSON, and require less storage space

```python
df.to_parquet("firstfile.parquet")
```

```python
pd.read_parquet("firstfile.parquet")
```
