# Aim: To perform Naïve Bayes and K-NN (k-nearest neighbor) classification using sklearn package in python

**Why is it called Naïve Bayes?**

# We have to create a Naïve Bayes model to classify flowers into one of the three given possible classes

```python
# Import libraries
import pandas as pd
import numpy as np
import sklearn.datasets as data
```

```python
# Read or Generate the dataset for machine learning task
d1 = data.load_iris()
d1
```

```python
# PLACE THE IRIS DATA IN A PANDAS 
# DATAFRAME 
iris = pd.DataFrame(data=d1.data,columns=d1.feature_names) 
iris
```

```python
# Add the target variable to the dataframe
iris['target'] = d1.target
```

```python
iris
```

```python
# Arrange data into a features matrix (X) and target vector (y)
# Create the Predictor and Target sets
X= d1.data
y= d1.target
```

```python
X
```

```python
# Splitting the data into train and test
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.30, random_state = 99)
```

```python
# Choose a class of model by importing the appropriate estimator class from Scikit-Learn
from sklearn.naive_bayes import GaussianNB
nv= GaussianNB()
```

```python
# Fit the model to your training data by calling the fit() method of the model instance
nv.fit(X_train,y_train)
```

```python
# Test the classifier on Test data
y_pred = nv.predict(X_test) # y_pred is predicted output
```

```python
# Compare predicted output with actual output to find accuracy
from sklearn.metrics import classification_report, confusion_matrix, accuracy_score
print(accuracy_score(y_test,y_pred))
print(confusion_matrix(y_test,y_pred))
print(classification_report(y_test, y_pred))
```

```python
new_data=[[6.5,3.1,1.2,1.5]]
```

```python
# Predict target values

predicted_target = nv.predict(new_data) 
print(predicted_target) 
```

# K-NN (k-nearest neighbor) classification 

```python
# Choose a class of model by importing the appropriate estimator class from Scikit-Learn
from sklearn.neighbors import KNeighborsClassifier
knn = KNeighborsClassifier(n_neighbors=3)
```

```python
# Fit the model to your training data by calling the fit() method of the model instance
knn.fit(X_train,y_train)
```

```python
# Test the classifier on Test data
y_pred = knn.predict(X_test) # y_pred is predicted output
```

```python
# Compare predicted output with actual output to find accuracy
from sklearn.metrics import classification_report, confusion_matrix, accuracy_score
print(accuracy_score(y_test,y_pred))
print(confusion_matrix(y_test,y_pred))
print(classification_report(y_test, y_pred))
```

```python
#Using drop function to exclude collums I won't need. The parameter inplace true will change the table 
data.drop(["PassengerId", "Name","SibSp","Parch","Ticket","Fare","Cabin","Embarked"], axis="columns",inplace=True)
data.head()
```

```python
X = data.drop(["Survived"],axis="columns")
# For the variable Y I can simply select the column Survived 
y = data["Survived"] #Another way to declare: Y = data.Survived 
y
```

```python
#The next step: create a pattern
#Pattern: male = 0 / female = 1
#This method map will create synonims in a dictionary
X.Sex = X.Sex.map({"male":0,"female":1})
X.head()
```

```python
#Some filles with the ages are empy. Checkin out the first 40 lines
X.Age[0:40]
```

```python
# use when someone used dataset from kaggle

# To fix this, we will use method fillna(). Inside the brackets I'll put the information to replace, which will be the mean
X.Age = X.Age.fillna(X.Age.mean())
X.Age[0:40]
```

```python
#Now want data to training (80%) and testing (20%)
train_test_split(X,y,train_size = 0.8) # It is automaticaly setting 0.2 for testing
# X_training [712 rows x 3 columns]
# X_test [179 rows x 3 columns]
# Y_training[Length: 712]
# Y_test[Length: 179]
```

```python
# Now create 4 variables to place each parameter
X_train,X_test,y_train,y_test=train_test_split(X,y,train_size = 0.8)
```

```python
# Choose a class of model by importing the appropriate estimator class from Scikit-Learn
from sklearn.naive_bayes import GaussianNB
nv= GaussianNB()
```

```python
# Fit the model to your training data by calling the fit() method of the model instance
nv.fit(X_train,y_train)
```

```python
# Test the classifier on Test data
y_pred = nv.predict(X_test) # y_pred is predicted output
```

```python
# Compare predicted output with actual output to find accuracy
from sklearn.metrics import classification_report, confusion_matrix, accuracy_score
print(accuracy_score(y_test,y_pred))
print(confusion_matrix(y_test,y_pred))
print(classification_report(y_test, y_pred))
```
