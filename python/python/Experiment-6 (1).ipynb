# Aim: To perform Decision Tree Classification (DCT) using sklearn package in python

# We have to create a Decision Tree classifier model to classify flowers into one of the three given possible classes

```python
# Import libraries
import pandas as pd
import numpy as np
import sklearn.datasets as data
```

```python
# Read or Generate the dataset for machine learning task
d1 = data.load_iris()
d1
```

```python
# PLACE THE IRIS DATA IN A PANDAS 
# DATAFRAME 
iris = pd.DataFrame(data=d1.data,columns=d1.feature_names) 
iris
```

```python
# Add the target variable to the dataframe
iris['target'] = d1.target
```

```python
iris
```

```python
# Arrange data into a features matrix (X) and target vector (y)
# Create the Predictor and Target sets
X= d1.data
y= d1.target
```

```python
# Splitting the data into train and test
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.30, random_state = 99)
```

```python
# Choose a class of model by importing the appropriate estimator class from Scikit-Learn
from sklearn.tree import DecisionTreeClassifier
dct = DecisionTreeClassifier()
```

```python
# Fit the model to your training data by calling the fit() method of the model instance
dct.fit(X_train,y_train)
```

```python
# Test the classifier on Test data
y_pred = dct.predict(X_test) # y_pred is predicted output
```

```python
# Compare predicted output with actual output to find accuracy
from sklearn.metrics import classification_report, confusion_matrix, accuracy_score
print(accuracy_score(y_test,y_pred))
print(confusion_matrix(y_test,y_pred))
print(classification_report(y_test, y_pred))
```

```python
# Visualizing Decision Tree

# Import necessary libraries for graph viz 
import matplotlib.pyplot as plt
from sklearn import tree
```

```python
fig, axes = plt.subplots( figsize = (4,6), dpi = 300)

tree.plot_tree(dct,filled=True)
fig.savefig('dct.png')
```

```python
# Let's do the same thing with titanic dataset
# Import nessasary libraries
import pandas as pd 
from sklearn.tree import plot_tree
import matplotlib.pyplot as plt
from sklearn import tree
from sklearn.model_selection import train_test_split
```

```python
data=pd.read_csv("train.csv")
```

```python
data.head()
```

```python
#Using drop function to exclude collums I won't need. The parameter inplace true will change the table 
data.drop(["PassengerId", "Name","SibSp","Parch","Ticket","Fare","Cabin","Embarked"], axis="columns",inplace=True)
data.head()
```

```python
X = data.drop(["Survived"],axis="columns")
# For the variable Y I can simply select the column Survived 
Y = data["Survived"] #Another way to declare: Y = data.Survived 
Y
```

```python
#The next step: create a pattern
#Pattern: male = 0 / female = 1
#This method map will create synonims in a dictionary
X.Sex = X.Sex.map({"male":0,"female":1})
X.head()
```

```python
#Some filles with the ages are empy. Checkin out the first 40 lines
X.Age[0:40]
```

```python
# use when someone used dataset from kaggle

# To fix this, we will use method fillna(). Inside the brackets I'll put the information to replace, which will be the mean
X.Age = X.Age.fillna(X.Age.mean())
X.Age[0:40]
```

```python
#Now want data to training (80%) and testing (20%)
train_test_split(X,Y,train_size = 0.8) # It is automaticaly setting 0.2 for testing
# X_training [712 rows x 3 columns]
# X_test [179 rows x 3 columns]
# Y_training[Length: 712]
# Y_test[Length: 179]
```

```python
# Now create 4 variables to place each parameter
X_train,X_test,Y_train,Y_test=train_test_split(X,Y,train_size = 0.8)
```

```python
len(X_train)
```

```python
len(X_test)
```

```python
X_test
# X_test is not in order and presents it ramdomly to prevent model from becoming biased 
```

```python
#If it is empty it will create it with as many levels it can. 
#TO avoid this we put max_dep to determine how many levels we want
model_dt = tree.DecisionTreeClassifier(max_depth=3)   #model with deph 3
model_dt
```

```python
#The method fit will adjust data for trainning the model
model_dt.fit(X_train,Y_train)
```

```python
#Using predict method to test the model
model_dt.predict(X_test)
```

```python
#Store the result in a variable 
Y_pred= model_dt.predict(X_test)
Y_pred
```

```python
#Dataframe to show Y real and Y predicted:
#Using a dictionary to separate columns  
result = pd.DataFrame({"Survided Actual": Y_test, "Survived Predicted": Y_pred})
result
```

```python
# Plotting the tree
plt.figure(figsize=(40,40))
graph_tree = plot_tree (model_dt,feature_names = ['Pclass','Sex','Age'],
                        class_names = ['Survived','Not Survived'],
                        filled=True,rounded=True,fontsize = 20)
```

```python
from sklearn import metrics
accuracy = metrics.accuracy_score(Y_test,Y_pred)  
accuracy
```

```python
precision= metrics.precision_score(Y_test,Y_pred)  
precision
```

```python
recall = metrics.recall_score(Y_test,Y_pred)  
recall
```

```python
confusion_matrix = metrics.confusion_matrix(Y_test,Y_pred)
confusion_matrix
```

```python
import seaborn as sns
sns.heatmap(confusion_matrix, annot=True)

# 0,0: 53 => True Negative
# 0,1: 0  => False Positive 
# 1,0: 23 => False Negative 
# 1,1: 4  => True Positive 
```
